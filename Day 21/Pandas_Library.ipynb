```python
#Series Real Example
Stock_prices = pd.Series([157,156,143,123,453], index=['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday'])
print(Stock_prices['Tuesday'])
```

```python
# Calculate the Mean 
mean_price = Stock_prices.mean()
print("Mean Stock Price:", mean_price)

#Calculate the Median
median_price = Stock_prices.median()
print("Median Stock Price:", median_price)

#Calculate the Max Value
max_price = Stock_prices.max()
print("Max Stock Price:", max_price)
```

```python
print(Stock_prices.describe())
print(Stock_prices.value_counts())
print(Stock_prices[Stock_prices > 150])
```

```python
# Dateframe with tow dimensional saa with the rows and columns
df = pd.DataFrame(data)


data={
    'Name': ['Kritam', 'Manish', 'Khagen'],
    'Age': [22, 23, 24],
    'Address': ['Kathmandu', 'Lalitpur', 'Bhaktapur']
}
```

```python
df = pd.DataFrame(data)
print(df)
```

```python
df.head(3)
```

```python
df.tail(1)
```

```python
# column Name
df['Name']
```

```python
df.shape
```

```python
df['Name']
```

```python
#Filtering Dataframe
filtered_df = df[df['Age'] > 22]
print(filtered_df)
```

```python
#Multiple Condition
filtered_df = df[(df['Age'] > 22) & (df['Address'] == 'Lalitpur')]
print(filtered_df)
```

```python
#Or condition
filtered_df = df[(df['Age'] > 22) | (df['Address'] == 'Lalitpur')]
print(filtered_df)
```

```python
#Add Upadate Delete 
#Adding new Column
df['Salary'] = [50000, 60000, 55000]
print(df)
```

```python
# Updating the existing column with Conditions
df.loc[df['Name'] == 'Kritam', 'Salary'] = 52000
print(df)
```

```python
#Line Plot
import matplotlib.pyplot as plt
```

```python


days = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']
temperatures = [30, 32, 31, 33, 32, 34, 33]

plt.plot(days, temperatures, marker='o', label='Temperatures')

plt.title('Temperature over a week')
plt.xlabel('Days')
plt.ylabel('Temperature (C)')
plt.legend() 
plt.show()
```

```python
#Bar Chart
products = ['Mobile', 'Laptop', 'Tablet']
sales =[ 100, 150, 80]
plt.bar(products, sales, color=['blue', 'green', 'yellow'])
plt.title('Sales of Products')
plt.xlabel('Products')
plt.ylabel('Sales')
plt.show()

```

```python
# Histogram
ages = [22, 25, 27, 30, 35, 40, 45, 50, 55, 60]
plt.hist(ages, bins=5, color='skyblue', edgecolor='black')
plt.title('Age Distribution')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.show()

    
```

```python
#Pie Chart
labels = ['Python', 'JavaScript', 'Java', 'C++']    
sizes = [40, 30, 20, 10]
plt.pie(sizes, labels=labels, autopct='%1.1f%%', startangle=140)
plt.title('Programming Language Popularity')
plt.show()

    
```
